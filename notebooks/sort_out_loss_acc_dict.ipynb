```python
import pickle

file_path = '../logs/combined_f_combined/resnet18_log_data.pkl'
test_path = '../logs/combined_f_combined/resnet18_test_data.pkl'
with open(file_path, 'rb') as f:
    result = pickle.load(f)

val_loss = result['vl']
val_acc = result['va']
new_val_loss = {}
new_val_acc = {}
```

```python
len(val_loss)
```

```python
with open(test_path, 'rb') as f:
    test = pickle.load(f)

test_loss = test['tel']
test_loss
```

```python
rekey_list = ['er', 'f', 'r', 'fx', 'cr']
for i, key in enumerate(val_loss):
    new_val_loss[rekey_list[i]] = val_loss[key]
    new_val_acc[rekey_list[i]] = val_acc[key]
```

```python
test_loss = {}
test_acc = {}

test_loss['cr'] = new_val_loss.pop('cr')
test_acc['cr'] = new_val_acc.pop('cr')
```

```python
new_val = {'tl': result['tl'], 'ta': result['ta'], 'vl': new_val_loss, 'va': new_val_acc}
result_test = {'tel': test_loss, 'tea': test_acc}
with open(file_path, 'wb') as f:
    pickle.dump(new_val, f)
with open(test_path, 'wb') as f:
    pickle.dump(result_test, f)
```
