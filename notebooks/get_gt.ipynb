```python
import shutil
import os

view = 'fx'
content_folder = f'../data/robustness/ood_{view}/content'
content_images = set(os.listdir(content_folder))

gt_folders = [f'../data/gt_ood_meadow_{view}_f',
              f'../data/gt_ood_forest_{view}_f',
              f'../data/gt_ood_desert_{view}_f',
              f'../data/gt_ood_industrial_{view}_f']

output_gt_folder = f'../data/robustness/ood_{view}/gt'

if not os.path.exists(output_gt_folder):
    os.makedirs(output_gt_folder)

for gt_folder in gt_folders:
    for image_name in os.listdir(gt_folder):
        if image_name in content_images:
            source_image_path = os.path.join(gt_folder, image_name)
            destination_image_path = os.path.join(output_gt_folder, image_name)
            shutil.copy(source_image_path, destination_image_path)

print(f"Images copied to {output_gt_folder} successfully.")
```
