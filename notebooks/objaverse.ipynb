```python
import json

with open('../objaverse/parsed_lvis_annotations.json', 'r') as f:
    lvis_data = json.load(f)

categories = list(sorted(set(lvis_data.values())))
```

```python
with open('../objaverse/lvis_categories.json', 'w') as f:
    json.dump(categories, f)
```
